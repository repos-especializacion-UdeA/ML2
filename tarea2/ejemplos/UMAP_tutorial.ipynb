<h3>🔹 ¿Qué es UMAP?</h3>

<p><b>Uniform Manifold Approximation and Projection (UMAP)</b> es una técnica de reducción de dimensionalidad que busca preservar la estructura de los datos en un espacio de menor dimensión. Se basa en la construcción de un <b>grafo de vecinos</b> y una optimización que mantiene la conectividad.</p>

<hr>

<h3>🔹 Modelo</h3>

<p>El método <b>UMAP (Uniform Manifold Approximation and Projection)</b> es una técnica avanzada de reducción de dimensionalidad que busca preservar tanto la estructura local como la global de los datos en espacios de menor dimensión. Su fundamentación matemática se basa en la teoría de variedades, grafos ponderados y optimización estocástica.</p>

<p>UMAP asume que los datos de alta dimensión están distribuidos en una variedad diferenciable y emplea un enfoque probabilístico para aprender una representación en un espacio de menor dimensión. Este proceso se lleva a cabo en dos etapas fundamentales:</p>

<ul>
  <li><b>Construcción de un grafo de vecinos:</b> Se modela la distribución de los datos en el espacio original mediante un grafo de proximidad, en el cual los nodos representan muestras y las aristas están ponderadas por una función de afinidad que mide la similitud entre puntos.</li>
  <li><b>Proyección en un espacio de menor dimensión:</b> Se encuentra una representación que mantiene la conectividad del grafo original, optimizando una función de pérdida que minimiza la diferencia entre la distribución en el espacio original y el espacio embebido.</li>
</ul>

<p>La optimización en UMAP se basa en minimizar la siguiente función de costo, que modela la conservación de la estructura topológica entre los espacios de alta y baja dimensión:</p>

$$
\mathcal{L} = \sum_{i,j} \left[ p_{ij} \log \frac{p_{ij}}{q_{ij}} + (1 - p_{ij}) \log \frac{1 - p_{ij}}{1 - q_{ij}} \right]
$$

<p>donde:</p>

<ul>
  <li>\( p_{ij} \) representa la probabilidad de conexión entre los puntos \( i \) y \( j \) en el espacio original. Esta probabilidad se define a partir de una distribución basada en un grafo de vecinos con pesos adaptativos.</li>
  <li>\( q_{ij} \) es la probabilidad correspondiente en el espacio embebido, calculada a partir de una distribución de Bernoulli que modela la proximidad relativa en la nueva representación.</li>
</ul>

<h3>📌 Relación con Otras Técnicas de Reducción de Dimensionalidad</h3>

<p>En comparación con métodos tradicionales como <b>PCA (Principal Component Analysis)</b> y <b>t-SNE (t-Distributed Stochastic Neighbor Embedding)</b>, UMAP presenta varias ventajas teóricas y computacionales:</p>

<ul>
  <li><b>PCA:</b> Se basa en la descomposición espectral de la matriz de covarianza y captura únicamente relaciones lineales entre las variables. UMAP, en contraste, preserva relaciones no lineales y estructuras más complejas en los datos.</li>
  <li><b>t-SNE:</b> Aunque es eficiente en la conservación de relaciones locales, t-SNE no mantiene bien la estructura global y tiende a ser computacionalmente más costoso en comparación con UMAP.</li>
</ul>

<h3>💡 Importancia de los Parámetros en UMAP</h3>

<p>Los resultados del embebimiento dependen de la configuración de parámetros como:</p>

<ul>
  <li><b><code>n_neighbors</code></b>: Controla el balance entre la preservación de la estructura local y global.</li>
  <li><b><code>min_dist</code></b>: Determina la compactación de los clusters en el espacio embebido.</li>
  <li><b><code>metric</code></b>: Afecta la medida de similitud utilizada en la construcción del grafo de vecinos.</li>
</ul>

<p>Se recomienda realizar ajustes iterativos y validaciones visuales para optimizar el embebimiento de acuerdo con las características específicas del conjunto de datos.</p>

<h2 align="center">📌 Parámetros de UMAP y su Significado</h2>

<hr>

<h3>🔹 1. <code>n_neighbors</code> (Número de Vecinos)</h3>

<p>Este parámetro determina el número de vecinos más cercanos que UMAP considerará al aprender la estructura de la variedad de los datos.</p>

<ul>
  <li><b>Valores bajos</b> (ej. 2 o 5) → UMAP se enfoca en la <b>estructura local</b>, capturando detalles finos pero perdiendo la perspectiva global.</li>
  <li><b>Valores altos</b> (ej. 50 o 100) → Se preserva más la <b>estructura global</b>, suavizando detalles locales pero capturando la forma general de los datos.</li>
</ul>

<hr>

<h3>🔹 2. <code>min_dist</code> (Distancia Mínima)</h3>

<p>Este parámetro controla la distancia mínima permitida entre puntos en el espacio embebido.</p>

<ul>
  <li><b>Valores bajos</b> → Los puntos se agrupan <b>más estrechamente</b>, resaltando la estructura local y posibles clústeres.</li>
  <li><b>Valores altos</b> → Mantienen una <b>mayor separación</b> entre los puntos, preservando la estructura global y evitando clústeres demasiado compactos.</li>
</ul>

<hr>

<h3>🔹 3. <code>n_components</code> (Dimensión del Espacio Embebido)</h3>

```python
# 📌 Instalar UMAP si no está instalado
!pip install umap-learn

# 📌 Importar librerías necesarias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import umap
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler
from mpl_toolkits.mplot3d import Axes3D

# 📌 Cargar el conjunto de datos (Dígitos escritos a mano)
digits = load_digits()
X = digits.data
y = digits.target

# 📌 Estandarizar los datos
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 📌 Aplicar PCA para comparación
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# 📌 Aplicar t-SNE para comparación
tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X_scaled)

# 📌 Aplicar UMAP
umap_model = umap.UMAP(n_components=2, random_state=42,)
X_umap = umap_model.fit_transform(X_scaled)

# 📌 Función para graficar los embebimientos
def plot_embedding(X_embedded, y, title):
    plt.figure(figsize=(8, 6))
    scatter = plt.scatter(X_embedded[:, 0], X_embedded[:, 1], c=y, cmap="Spectral", alpha=0.7)
    plt.colorbar(scatter, label="Dígito")
    plt.title(title, fontsize=14)
    plt.xlabel("Componente 1")
    plt.ylabel("Componente 2")
    plt.show()

# 📌 Comparar visualmente PCA, t-SNE y UMAP
plot_embedding(X_pca, y, "Embebimiento usando PCA")
plot_embedding(X_tsne, y, "Embebimiento usando t-SNE")
plot_embedding(X_umap, y, "Embebimiento usando UMAP")

# 📌 Graficar en 3D con UMAP
umap_model_3d = umap.UMAP(n_components=3, random_state=42)
X_umap_3d = umap_model_3d.fit_transform(X_scaled)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(X_umap_3d[:, 0], X_umap_3d[:, 1], X_umap_3d[:, 2], c=y, cmap="Spectral", alpha=0.7)
ax.set_title("Embebimiento 3D con UMAP")
ax.set_xlabel("Componente 1")
ax.set_ylabel("Componente 2")
ax.set_zlabel("Componente 3")
plt.colorbar(scatter, label="Dígito")
plt.show()
```
